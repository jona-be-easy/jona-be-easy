```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("Zomato Restaurant Dataset.csv")

data.head()
```

```python
data.columns
```

```python
data.plot.hist(y="Aggregate rating",edgecolor="black", density=True,title="Histogram")
data["Aggregate rating"].plot.kde(color="red")

mean_value=data["Aggregate rating"].mean()
median_value=data["Aggregate rating"].median()
modal_value=data["Aggregate rating"].mode()[0]
plt.axvline(mean_value,color="green", label=f"mean value:{mean_value:.1f}")
plt.axvline(median_value,color="gold", label=f"median value:{median_value:.1f}")
plt.axvline(modal_value, color="purple", label=f"modal value:{modal_value:.1f}")


plt.legend()
plt.show()
```

The plot shows bimodal distribution of the user ratings with two distinct peaks
The 0.0 peak shows a sharp spike at exactly 0 representing new or unrated items.
the 2 to 5 peak shows a natural bell curve representing the genuine ratings of the active reviewed items.
the curve is not skewed as well.

```python
data.plot.scatter(y="Votes",x="Aggregate rating")
```

- The scatter plot shows a positive correlation between an item's aggregate rating and it's number of votes
- The 0.0 cluster shows that items rated 0.0 have virtually zero votes, proving they are unrated or new items
- the upward funnel shows that as ratings incraese, the number of votes climbs drastically. popular items with high engagement consistently mainatin high scores

```python

data.plot.scatter(y="Aggregate rating",x="Average Cost for two",xlabel="Average Cost for two",ylabel="Aggregate ratings",figsize=(13,5),title="A SCATTER DIAGRAM SHOWING RELATIONSHIP OF AGGREGATE RATINGS WITH AVERAGE COST FOR TWO ")
plt.show()
```

- The scatter plot shows that the average cost for two positively correlates with the average ratings
- This shows that some restaurants with higher average cost for two  tend to have a higher aggregate rating

```python
# Box plots are good for  comparing different columns because they summarize the distribution of numerical data by showing  median, quartiles and any outliers
sns.boxplot(data=data,y="Aggregate rating",x="Price range")
plt.show()
```

- Shows that restaurants in the higher price ranges tend to receive a higher median aggregate rating
- The outliers in the higher price ratings show that a higher price range doesnt guarantee a good rating

```python
sns.boxplot(data=data,y="Aggregate rating",x="Has Table booking")
plt.show()
```

- Restarants offering table booking tend to have higher median aggregate ratings compared to those that do not offer table booking.
- Most restaurants offering table booking get higher aggregate ratings compared to those that don't have.
- We should however note that there are some restaurants thatoffer table booking and have low table ratings.

```python
sns.boxplot(data=data,y="Aggregate rating",x="Has Online delivery")
plt.show()
```
